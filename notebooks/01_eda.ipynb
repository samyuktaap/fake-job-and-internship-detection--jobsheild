# JobShield: 01 - Exploratory Data Analysis (EDA)
## Dataset Research, Hygiene, and Class Imbalance Inspection

This notebook inspects the Employment Scam Aegean Dataset (EMSCAD / Kaggle Fake Job Posting Prediction dataset).

```python
import pandas as pd
import numpy as np

# Load dataset
df = pd.read_csv('../data/fake_job_postings.csv')
print('Raw Shape:', df.shape)
df.head(3)
```

### 1. Label Distribution & Class Imbalance
In real-world hiring environments, fraudulent listings constitute a tiny fraction of total job postings. Identifying this baseline imbalance is critical to selecting proper evaluation metrics (PR-AUC, Recall, Precision instead of naive accuracy).

```python
counts = df['fraudulent'].value_counts()
proportions = df['fraudulent'].value_counts(normalize=True) * 100
print('Counts:
', counts)
print('Proportions (%):
', proportions)
```

### 2. Missing Value Analysis
Missingness in specific fields (such as `company_profile` and `has_company_logo`) often serves as a structural signal of unverified employers.

```python
missing = df.isna().sum()
missing_pct = (df.isna().sum() / len(df)) * 100
missing_df = pd.DataFrame({'Missing Count': missing, 'Missing %': missing_pct})
print(missing_df[missing_df['Missing Count'] > 0].sort_values(by='Missing %', ascending=False))
```

### 3. Logo and Question Presence vs Fraud Rate

```python
print('Fraud Rate with Logo:', df[df['has_company_logo'] == 1]['fraudulent'].mean() * 100)
print('Fraud Rate WITHOUT Logo:', df[df['has_company_logo'] == 0]['fraudulent'].mean() * 100)
```

### Summary of EDA Findings:
- Severe class imbalance: only ~4.8% of raw listings are fraudulent.
- Over 2,000 duplicate job templates exist; dropping them is necessary to prevent data leakage between train and test sets.
- Postings without company logos have a significantly higher fraud rate.
